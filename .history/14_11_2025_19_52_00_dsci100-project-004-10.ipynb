## 2. Research Question

### Broad Question
**Question 2:** We would like to know which "kinds" of players are most likely to contribute a large amount of data so that we can target those players in our recruiting efforts.

### Specific Research Question
**Can player characteristics (experience level, age, gender, and subscription status) predict the total number of gameplay sessions contributed by a player in the Minecraft research server dataset?**

**Response Variable:** Number of sessions per player (count data, derived from sessions.csv)

**Explanatory Variables:**
- Experience level (categorical: Beginner, Amateur, Regular, Veteran, Pro)
- Age (numeric, in years)
- Gender (categorical)
- Newsletter subscription status (binary: TRUE/FALSE)

### How the Data Addresses the Question

The datasets enable this analysis by:
1. Linking players to sessions via hashedEmail to calculate session counts per player
2. Providing demographic and behavioral characteristics that may predict engagement
3. Measuring data contribution through session counts as a proxy for research value
4. Supporting targeted recruitment by identifying high-value player profiles

## 3. Exploratory Data Analysis and Visualization

### Data Wrangling

```r
# Count sessions per player
session_counts <- sessions %>%
  filter(!is.na(end_time)) %>% 
  group_by(hashedEmail) %>%
  summarise(num_sessions = n())

# Join with player data
analysis_data <- players %>%
  left_join(session_counts, by = "hashedEmail") %>%
  mutate(
    num_sessions = replace_na(num_sessions, 0),
    Age = as.numeric(Age),  
    experience = factor(experience, 
                       levels = c("Beginner", "Amateur", "Regular", "Veteran", "Pro"),
                       ordered = TRUE)
  )

cat("Analysis dataset created with", nrow(analysis_data), "players\n")
cat("Players with at least one session:", sum(analysis_data$num_sessions > 0), "\n")
cat("Average sessions per player:", round(mean(analysis_data$num_sessions), 2), "\n")
```

### Visualization 1: Do Experience Levels Differ in Engagement?
I want to see if more experienced players have more sessions than less experienced players.

```r
ggplot(analysis_data, aes(x = experience, y = num_sessions, fill = experience)) +
  geom_boxplot() +
  labs(
    title = "Number of Sessions by Player Experience Level",
    x = "Experience Level",
    y = "Number of Sessions",
    caption = "Figure 1: Boxplot showing how session counts vary by experience level"
  )
```

**Insight:** Experience level shows clear differences in engagement patterns. Pro and Veteran players tend to have more sessions than Beginners, though there's substantial variation within each group. This suggests experience level will be an important predictor in our model.

### Visualization 2: Age vs Session Count

```r
ggplot(analysis_data %>% filter(!is.na(Age)), 
       aes(x = Age, y = num_sessions)) +
  geom_point(alpha = 0.3, color = "steelblue", size = 2) +
  geom_smooth(method = "loess", se = TRUE, color = "darkred", linewidth = 1) +
  labs(
    title = "Relationship Between Player Age and Session Count",
    x = "Age (years)",
    y = "Number of Sessions",
    caption = "Figure 2: Scatterplot with LOESS smoothing curve showing trend.\nEach point represents one player."
  ) +
  theme_minimal(base_size = 12)
```

**Insight:** The relationship between age and session count appears non-linear, with younger players (teens to early 20s) showing the highest engagement. There's considerable variability across all age groups, suggesting age alone may not be a strong predictor, but could interact with other variables.

### Visualization 3: Subscription Status and Engagement

```r
ggplot(analysis_data, aes(x = subscribe, y = num_sessions + 1, fill = subscribe)) +
  geom_violin(alpha = 0.7, trim = FALSE) +
  geom_boxplot(width = 0.2, alpha = 0.8, outlier.alpha = 0.3) +
  scale_y_log10(labels = scales::comma) +
  labs(
    title = "Session Counts: Subscribed vs Non-Subscribed Players",
    x = "Newsletter Subscription Status",
    y = "Number of Sessions (log scale)",
    caption = "Figure 3: Violin and box plots comparing engagement by subscription status.\nAdded 1 to all values to handle zeros on log scale."
  ) +
  scale_fill_manual(values = c("FALSE" = "#E74C3C", "TRUE" = "#2ECC71")) +
  theme_minimal(base_size = 12) +
  theme(legend.position = "none")
```

**Insight:** Newsletter subscribers show notably higher median session counts compared to non-subscribers. However, subscription status could be both a predictor and an indicator of existing engagement. This correlation suggests subscribers are more committed players, making this variable valuable for recruitment targeting.
